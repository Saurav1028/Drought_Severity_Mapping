```python
# ============================================================
# STEP 49: CHECK IMAGE ID MATCHING
# ============================================================

print("============================================================")
print("STEP 49: CHECKING IMAGE ID MATCHING")
print("============================================================")

def get_image_ids(folder):
    return {
        file.stem
        for file in folder.glob("*.tif")
    }

ndvi_ids = get_image_ids(ndvi_folder)
rgb_ids = get_image_ids(rgb_folder)
vv_ids = get_image_ids(sar_vv_folder)
vh_ids = get_image_ids(sar_vh_folder)

print("\n📊 Image counts:")
print("NDVI :", len(ndvi_ids))
print("RGB  :", len(rgb_ids))
print("VV   :", len(vv_ids))
print("VH   :", len(vh_ids))

# Find IDs common to all modalities
common_ids = ndvi_ids & rgb_ids & vv_ids & vh_ids

print("\n🔗 Common image IDs:", len(common_ids))

# Find missing IDs
print("\n❓ Missing matches:")

print("NDVI missing:", len(common_ids - ndvi_ids))
print("RGB missing :", len(common_ids - rgb_ids))
print("VV missing  :", len(common_ids - vv_ids))
print("VH missing  :", len(common_ids - vh_ids))

# Show examples
print("\n📋 First 20 common IDs:")

for image_id in sorted(common_ids)[:20]:
    print("-", image_id)

print("\n============================================================")
print("✅ STEP 49 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 50: CHECK IMAGE DIMENSIONS
# ============================================================

print("============================================================")
print("STEP 50: CHECKING IMAGE DIMENSIONS")
print("============================================================")

def get_dimensions(folder):
    dimensions = {}

    for file in folder.glob("*.tif"):
        with rasterio.open(file) as src:
            dimensions[file.stem] = (src.width, src.height)

    return dimensions


print("\n📡 Checking NDVI dimensions...")
ndvi_dimensions = get_dimensions(ndvi_folder)

print("📡 Checking RGB dimensions...")
rgb_dimensions = get_dimensions(rgb_folder)

print("📡 Checking SAR VV dimensions...")
vv_dimensions = get_dimensions(sar_vv_folder)

print("📡 Checking SAR VH dimensions...")
vh_dimensions = get_dimensions(sar_vh_folder)


# Find unique dimensions
ndvi_unique = set(ndvi_dimensions.values())
rgb_unique = set(rgb_dimensions.values())
vv_unique = set(vv_dimensions.values())
vh_unique = set(vh_dimensions.values())


print("\n============================================================")
print("📊 UNIQUE IMAGE DIMENSIONS")
print("============================================================")

print("\nNDVI:", ndvi_unique)
print("RGB :", rgb_unique)
print("VV  :", vv_unique)
print("VH  :", vh_unique)


# Check whether dimensions are identical across modalities
all_ids = sorted(common_ids)

dimension_mismatches = []

for image_id in all_ids:

    dimensions = {
        "NDVI": ndvi_dimensions[image_id],
        "RGB": rgb_dimensions[image_id],
        "VV": vv_dimensions[image_id],
        "VH": vh_dimensions[image_id]
    }

    if len(set(dimensions.values())) > 1:
        dimension_mismatches.append(
            (image_id, dimensions)
        )


print("\n============================================================")
print("🔎 DIMENSION MATCHING")
print("============================================================")

print(
    "Images with dimension mismatch:",
    len(dimension_mismatches)
)

if len(dimension_mismatches) > 0:

    print("\n⚠️ First 10 mismatches:")

    for item in dimension_mismatches[:10]:
        print(item)

else:

    print("✅ All modalities have matching dimensions.")


print("\n============================================================")
print("✅ STEP 50 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 51: CHECK IMAGE BAND STRUCTURE
# ============================================================

print("============================================================")
print("STEP 51: CHECKING IMAGE BAND STRUCTURE")
print("============================================================")

def get_band_counts(folder):
    band_counts = {}

    for file in folder.glob("*.tif"):
        with rasterio.open(file) as src:
            band_counts[file.stem] = src.count

    return band_counts


print("\n📡 Checking NDVI bands...")
ndvi_bands = get_band_counts(ndvi_folder)

print("📡 Checking RGB bands...")
rgb_bands = get_band_counts(rgb_folder)

print("📡 Checking SAR VV bands...")
vv_bands = get_band_counts(sar_vv_folder)

print("📡 Checking SAR VH bands...")
vh_bands = get_band_counts(sar_vh_folder)


# Unique band counts
ndvi_unique_bands = set(ndvi_bands.values())
rgb_unique_bands = set(rgb_bands.values())
vv_unique_bands = set(vv_bands.values())
vh_unique_bands = set(vh_bands.values())


print("\n============================================================")
print("📊 UNIQUE BAND COUNTS")
print("============================================================")

print("NDVI:", ndvi_unique_bands)
print("RGB :", rgb_unique_bands)
print("VV  :", vv_unique_bands)
print("VH  :", vh_unique_bands)


# Check consistency
print("\n============================================================")
print("🔎 BAND CONSISTENCY")
print("============================================================")

band_mismatches = []

for image_id in sorted(common_ids):

    bands = {
        "NDVI": ndvi_bands[image_id],
        "RGB": rgb_bands[image_id],
        "VV": vv_bands[image_id],
        "VH": vh_bands[image_id]
    }

    # Expected:
    # NDVI = 1
    # RGB  = 3
    # VV   = 1
    # VH   = 1

    expected = {
        "NDVI": 1,
        "RGB": 3,
        "VV": 1,
        "VH": 1
    }

    if bands != expected:
        band_mismatches.append(
            (image_id, bands)
        )


print(
    "Images with unexpected band structure:",
    len(band_mismatches)
)

if len(band_mismatches) > 0:

    print("\n⚠️ First 10 mismatches:")

    for item in band_mismatches[:10]:
        print(item)

else:

    print("✅ All 2,200 images have the expected band structure.")


print("\n============================================================")
print("✅ STEP 51 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 53: IMAGE PIXEL VALUE ANALYSIS
# ============================================================

print("============================================================")
print("STEP 53: IMAGE PIXEL VALUE ANALYSIS")
print("============================================================")

import numpy as np
import rasterio

# Number of images to sample
SAMPLE_SIZE = 20

def analyze_pixel_values(folder, sample_size=20):

    files = sorted(folder.glob("*.tif"))[:sample_size]

    minimums = []
    maximums = []
    means = []

    for file in files:

        with rasterio.open(file) as src:
            image = src.read().astype(np.float32)

        minimums.append(float(np.min(image)))
        maximums.append(float(np.max(image)))
        means.append(float(np.mean(image)))

    return {
        "images_analyzed": len(files),
        "minimum_pixel": min(minimums),
        "maximum_pixel": max(maximums),
        "mean_of_image_means": np.mean(means)
    }


# Analyze each modality
results = {
    "NDVI": analyze_pixel_values(ndvi_folder, SAMPLE_SIZE),
    "RGB": analyze_pixel_values(rgb_folder, SAMPLE_SIZE),
    "SAR_VV": analyze_pixel_values(sar_vv_folder, SAMPLE_SIZE),
    "SAR_VH": analyze_pixel_values(sar_vh_folder, SAMPLE_SIZE)
}


print("\n📊 PIXEL VALUE SUMMARY")
print("============================================================")

for modality, result in results.items():

    print(f"\n📡 {modality}")
    print(f"Images analyzed       : {result['images_analyzed']}")
    print(f"Minimum pixel value   : {result['minimum_pixel']:.4f}")
    print(f"Maximum pixel value   : {result['maximum_pixel']:.4f}")
    print(f"Mean image value      : {result['mean_of_image_means']:.4f}")


print("\n============================================================")
print("✅ STEP 53 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 54: NDVI DISTRIBUTION ANALYSIS
# ============================================================

print("============================================================")
print("STEP 54: NDVI DISTRIBUTION ANALYSIS")
print("============================================================")

# Sample 20 NDVI images
ndvi_files = sorted(ndvi_folder.glob("*.tif"))[:20]

all_ndvi_values = []

for file in ndvi_files:

    with rasterio.open(file) as src:
        image = src.read(1).astype(np.float32)

    # Remove invalid NaN/Inf values
    valid_values = image[np.isfinite(image)]

    all_ndvi_values.append(valid_values)

# Combine sampled pixels
all_ndvi_values = np.concatenate(all_ndvi_values)

print("\n📊 NDVI statistics from sampled images:")
print("Number of pixels:", len(all_ndvi_values))
print("Minimum:", np.min(all_ndvi_values))
print("Maximum:", np.max(all_ndvi_values))
print("Mean:", np.mean(all_ndvi_values))
print("Median:", np.median(all_ndvi_values))
print("Standard deviation:", np.std(all_ndvi_values))

# Plot distribution
plt.figure(figsize=(10, 6))

plt.hist(
    all_ndvi_values,
    bins=50,
    edgecolor="black"
)

plt.title("NDVI Pixel Value Distribution")
plt.xlabel("NDVI Value")
plt.ylabel("Pixel Frequency")
plt.grid(axis="y", alpha=0.3)

figure_path = (
    PROJECT_ROOT
    / "results"
    / "figures"
    / "ndvi_pixel_distribution.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\n📁 Figure saved at:")
print(figure_path)

print("\n============================================================")
print("✅ STEP 54 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 75: CHECK METADATA YEAR/MONTH CONSISTENCY
# ============================================================

print("============================================================")
print("STEP 75: CHECKING YEAR/MONTH CONSISTENCY")
print("============================================================")

metadata_check = df_processed.copy()

# Convert initial date
metadata_check["Initial_Date_Parsed"] = pd.to_datetime(
    metadata_check["Initial dates"],
    errors="coerce"
)

# Extract year and month from the date
metadata_check["Calculated_Year"] = (
    metadata_check["Initial_Date_Parsed"].dt.year
)

metadata_check["Calculated_Month"] = (
    metadata_check["Initial_Date_Parsed"].dt.month
)

# Compare with existing columns
metadata_check["Year_Match"] = (
    metadata_check["Year"] == metadata_check["Calculated_Year"]
)

metadata_check["Month_Match"] = (
    metadata_check["Month"] == metadata_check["Calculated_Month"]
)

print("\n📊 Total records:")
print(len(metadata_check))

print("\n📅 Year matches:")
print(metadata_check["Year_Match"].sum())

print("❌ Year mismatches:")
print((~metadata_check["Year_Match"]).sum())

print("\n📅 Month matches:")
print(metadata_check["Month_Match"].sum())

print("❌ Month mismatches:")
print((~metadata_check["Month_Match"]).sum())

# Show mismatches
mismatches = metadata_check[
    (~metadata_check["Year_Match"]) |
    (~metadata_check["Month_Match"])
]

if len(mismatches) > 0:

    print("\n⚠️ First mismatches:")
    display(
        mismatches[
            [
                "Initial dates",
                "Year",
                "Calculated_Year",
                "Month",
                "Calculated_Month"
            ]
        ].head(20)
    )

else:

    print("\n✅ Year and Month are completely consistent with Initial dates.")

print("\n============================================================")
print("✅ STEP 75 COMPLETED")
print("============================================================")
```

```python
# ============================================================
# STEP 76: CHECK METADATA DURATION CONSISTENCY
# ============================================================

print("============================================================")
print("STEP 76: CHECKING DURATION CONSISTENCY")
print("============================================================")

duration_check = df_processed.copy()

# Convert dates
duration_check["Initial_Date_Parsed"] = pd.to_datetime(
    duration_check["Initial dates"],
    errors="coerce"
)

duration_check["End_Date_Parsed"] = pd.to_datetime(
    duration_check["End dates"],
    errors="coerce"
)

# Calculate duration
duration_check["Calculated_Duration"] = (
    duration_check["End_Date_Parsed"]
    - duration_check["Initial_Date_Parsed"]
).dt.days

# Compare with existing Duration_Days
duration_check["Duration_Match"] = (
    duration_check["Duration_Days"]
    == duration_check["Calculated_Duration"]
)

print("\n📊 Total records:")
print(len(duration_check))

print("\n✅ Duration matches:")
print(duration_check["Duration_Match"].sum())

print("\n❌ Duration mismatches:")
print((~duration_check["Duration_Match"]).sum())

# Show mismatches
mismatches = duration_check[
    ~duration_check["Duration_Match"]
]

if len(mismatches) > 0:

    print("\n⚠️ First 20 duration mismatches:")

    display(
        mismatches[
            [
                "Initial dates",
                "End dates",
                "Duration_Days",
                "Calculated_Duration"
            ]
        ].head(20)
    )

else:

    print("\n✅ All Duration_Days values are consistent with the dates.")

print("\n============================================================")
print("✅ STEP 76 COMPLETED")
print("============================================================")
```
